```python
from pathlib import Path
import pandas as pd

files = sorted(Path("../data/raw").glob("*.csv"))
for f in files:
    print(f.name)
```

```python
for i, f in enumerate(files):
    print(i, f.name)
```

```python
df = pd.read_csv(files[1], index_col=0, low_memory=False)

print(df.shape)
print("实验数:", df["clickability_test_id"].nunique())
print("期间:", df["created_at"].min(), "〜", df["created_at"].max())
```

```python
df["created_at"] = pd.to_datetime(df["created_at"], format="ISO8601")
in_window = df["created_at"].between("2013-06-25", "2014-01-10 23:59:59")

print("问题时期的实验数:", df.loc[in_window, "clickability_test_id"].nunique())
print("全部实验数:", df["clickability_test_id"].nunique())
```
